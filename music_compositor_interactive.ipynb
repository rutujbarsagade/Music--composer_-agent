# AI Music Compositor - Interactive UI

## Advanced Music Generation with LangGraph

This notebook provides an interactive interface for generating AI-powered musical compositions.

```python
# Install and import required libraries
import sys
from pathlib import Path

# Add project root to path
project_root = Path.cwd()
sys.path.insert(0, str(project_root))

# Import required modules
from typing import Dict
from datetime import datetime
import logging
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

# Import project modules
from src.config import config
from src.workflow import create_workflow
from src.validators import validate_composition_input
from src.audio import AudioPlayer
from src.exceptions import (    MusicGenerationError,
    AudioPlaybackError,
    InvalidInputError)

# Configure logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

print("✅ All modules loaded successfully")
```

```python
# Global state
last_result = None
audio_player = None
app = None

def initialize_system():
    """Initialize the music composition system"""
    global app, audio_player
    try:
        # Validate configuration
        config.validate()
        logger.info("Configuration validated")
        
        # Create workflow
        app = create_workflow()
        logger.info("Workflow created")
        
        # Initialize audio player
        audio_player = AudioPlayer()
        logger.info("Audio player initialized")
        
        return True
    except Exception as e:
        logger.error(f"Initialization failed: {e}")
        return False

# Initialize on load
if initialize_system():
    print("🎵 Music Compositor system ready!")
else:
    print("❌ Failed to initialize system. Check your configuration.")
```

```python
def create_interactive_ui():
    """Create the interactive UI for music composition"""
    
    global last_result, app, audio_player
    
    # UI Container
    ui_container = widgets.VBox()
    
    # Title
    title = widgets.HTML(
        "<h2 style='color: #2E86AB; text-align: center;'>🎵 AI Music Compositor 🎵</h2>"
    )
    
    # Description
    description = widgets.HTML(
        "<p style='text-align: center; color: #555; font-size: 14px;'>"
        "Generate unique musical compositions using AI and LangGraph. "
        "Customize every aspect of your composition."
        "</p>"
    )
    
    # Section: Input Parameters
    input_section = widgets.HTML("<h3 style='color: #A23B72; margin-top: 20px;'>🎹 Composition Parameters</h3>")
    
    # Music description
    music_description = widgets.Textarea(
        value="Create a happy piano piece in C major",
        placeholder="Describe the music you want to generate...",
        description="Description:",
        rows=3,
        layout=widgets.Layout(width='100%')
    )
    
    # Style dropdown
    musical_style = widgets.Dropdown(
        options=config.MUSICAL_STYLES,
        value="Romantic era",
        description="Style:",
        style={'description_width': 'initial'},
        layout=widgets.Layout(width='100%')
    )
    
    # Key dropdown
    key_selection = widgets.Dropdown(
        options=config.KEYS,
        value="C major",
        description="Key:",
        style={'description_width': 'initial'},
        layout=widgets.Layout(width='100%')
    )
    
    # Instrument dropdown
    instrument = widgets.Dropdown(
        options=config.INSTRUMENTS,
        value="Piano",
        description="Instrument:",
        style={'description_width': 'initial'},
        layout=widgets.Layout(width='100%')
    )
    
    # Tempo slider
    tempo = widgets.IntSlider(
        value=60,
        min=40,
        max=180,
        step=5,
        description="Tempo (BPM):",
        style={'description_width': 'initial'},
        layout=widgets.Layout(width='100%')
    )
    
    # Duration slider
    duration = widgets.IntSlider(
        value=8,
        min=4,
        max=32,
        step=2,
        description="Duration (beats):",
        style={'description_width': 'initial'},
        layout=widgets.Layout(width='100%')
    )
    
    # Section: Controls
    controls_section = widgets.HTML("<h3 style='color: #A23B72; margin-top: 20px;'>🎛️ Controls</h3>")
    
    # Buttons
    generate_button = widgets.Button(
        description='Generate Music 🎶',
        button_style='success',
        tooltip='Generate a new composition',
        layout=widgets.Layout(width='48%', height='45px', font_size='14px')
    )
    
    play_button = widgets.Button(
        description='Play Music 🔊',
        button_style='info',
        tooltip='Play the generated composition',
        layout=widgets.Layout(width='48%', height='45px', font_size='14px')
    )
    
    reset_button = widgets.Button(
        description='Reset ↻',
        button_style='warning',
        tooltip='Reset all parameters',
        layout=widgets.Layout(width='48%', height='45px', font_size='14px')
    )
    
    stop_button = widgets.Button(
        description='Stop ⏹️',
        button_style='danger',
        tooltip='Stop playback',
        layout=widgets.Layout(width='48%', height='45px', font_size='14px')
    )
    
    # Status and results output
    status_output = widgets.Output()
    results_output = widgets.Output()
    
    # Button handlers
    def on_generate_click(b):
        global last_result
        try:
            with status_output:
                clear_output()
                print("🎵 Generating composition..."
                print(f"   Description: {music_description.value}")
                print(f"   Style: {musical_style.value}")
                print(f"   Key: {key_selection.value}")
                print(f"   Instrument: {instrument.value}")
                print(f"   Tempo: {tempo.value} BPM")
                print(f"   Duration: {duration.value} beats")
                print("\n⏳ Processing...")
            
            # Prepare input
            inputs = {
                "musician_input": f"{music_description.value} using {instrument.value}",
                "style": musical_style.value,
                "key": key_selection.value,
                "instrument": instrument.value,
                "tempo": tempo.value,
                "duration": duration.value
            }
            
            # Validate
            validate_composition_input(inputs)
            
            # Generate
            last_result = app.invoke(inputs)
            
            with status_output:
                clear_output()
                print("✅ Composition generated successfully!")
                print(f"\n📁 MIDI file: {last_result['midi_file']}")
                print(f"⏰ Generated: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
            
            with results_output:
                clear_output()
                print("🎼 Composition Details:")
                print(f"\nMelody (preview):")
                print(last_result['melody'][:150] + "..."
                print(f"\nHarmony (preview):")
                print(last_result['harmony'][:150] + "..."
        
        except InvalidInputError as e:
            with status_output:
                clear_output()
                print(f"⚠️ Input Error: {str(e)}")
        except MusicGenerationError as e:
            with status_output:
                clear_output()
                print(f"❌ Generation Error: {str(e)}")
        except Exception as e:
            with status_output:
                clear_output()
                print(f"❌ Unexpected Error: {str(e)}")
    
    def on_play_click(b):
        if last_result is None:
            with status_output:
                clear_output()
                print("⚠️ Please generate a composition first!")
            return
        
        try:
            with status_output:
                clear_output()
                print("🔊 Playing composition...")
            
            audio_player.play(last_result['midi_file'])
            
            with status_output:
                clear_output()
                print("✅ Playback complete!")
        except AudioPlaybackError as e:
            with status_output:
                clear_output()
                print(f"❌ Playback Error: {str(e)}")
        except Exception as e:
            with status_output:
                clear_output()
                print(f"❌ Error: {str(e)}")
    
    def on_reset_click(b):
        music_description.value = "Create a happy piano piece in C major"
        musical_style.value = "Romantic era"
        key_selection.value = "C major"
        instrument.value = "Piano"
        tempo.value = 60
        duration.value = 8
        
        with status_output:
            clear_output()
            print("↻ Parameters reset to defaults")
        
        with results_output:
            clear_output()
    
    def on_stop_click(b):
        try:
            audio_player.stop()
            with status_output:
                clear_output()
                print("⏹️ Playback stopped")
        except Exception as e:
            with status_output:
                clear_output()
                print(f"❌ Error stopping playback: {str(e)}")
    
    # Attach handlers
    generate_button.on_click(on_generate_click)
    play_button.on_click(on_play_click)
    reset_button.on_click(on_reset_click)
    stop_button.on_click(on_stop_click)
    
    # Arrange UI
    button_row1 = widgets.HBox([generate_button, play_button])
    button_row2 = widgets.HBox([reset_button, stop_button])
    
    params_box = widgets.VBox([
        widgets.HTML("<p style='font-weight: bold;'>Description:</p>"),
        music_description,
        widgets.HTML("<p style='font-weight: bold; margin-top: 15px;'>Musical Settings:</p>"),
        musical_style,
        key_selection,
        instrument,
        widgets.HTML("<p style='font-weight: bold; margin-top: 15px;'>Playback Settings:</p>"),
        tempo,
        duration
    ])
    
    # Main UI
    ui = widgets.VBox([
        title,
        description,
        input_section,
        params_box,
        controls_section,
        button_row1,
        button_row2,
        widgets.HTML("<hr style='margin: 20px 0;'>"),
        status_output,
        widgets.HTML("<h3 style='color: #A23B72;'>📊 Results</h3>"),
        results_output
    ])
    
    return ui

# Display UI
print("Creating interactive UI...")
ui = create_interactive_ui()
display(ui)
print("✅ UI Ready!")
```
